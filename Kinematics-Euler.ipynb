```python
from vpython import *

# parametros del sistema
A = 10.0
B = 5.0
g = 9.8 #m/s^2

h = A**2/B**2*( 2/3*A + 2*g )

def a(t):
    return A - B*t

def _a(t):
    return -g

c = canvas(
  title='Simulación de cinemética',
  background = color.white
)

c.width = 500
c.height = 800

s = sphere(
  radius= 1.0,
  pos = vec(0,h,0),
  color = color.blue
)

rocket = cone(
  pos=vec(0, 0, 0),
  axis=vec(0, 2, 0),
  radius=1.0,
  color=color.yellow
)

# tiempo de colision
tmax = 2*A/B
t = [ (i+1)*(tmax)/500 for i in range(500) ]
dt = t[1]-t[0]

v_rocket = 0.0
y_rocket = 0.0

v_rock = 0.0
y_rock = s.pos.y

for ti in t:
    rate(30)

    # Euler para la velocidad
    v_rocket = v_rocket + a(ti)*dt

    # Euler para la posición
    y_rocket = y_rocket + v_rocket * dt

    rocket.pos.y = y_rocket

    # Euler para la velocidad
    v_rock = v_rock + _a(ti)*dt

    # Euler para la posición
    y_rock = y_rock + v_rock*dt

    s.pos.y = y_rock
```
