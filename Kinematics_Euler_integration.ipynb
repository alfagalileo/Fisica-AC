```python
from vpython import *
import numpy as np
```

```python
# parametros del sistema
A = 10.0 
B = 5.0
g = 9.8 #m/s^2

h = A**2/B**2*( 2/3*A + 2*g )

def a(t):
    return A - B*t

def _a(t):
    return -g

c = canvas(
    title='Simulación de cinemática',
    background = color.black,
    align="left"
)

c.width = 500
c.height = 800

s = sphere(
  radius= 2.0,
  pos = vec(0,h,0),
  color = color.blue
)

rocket = cone(
  pos=vec(0, 0, 0), 
  axis=vec(0, 4, 0), 
  radius=2.0, 
  color=color.yellow
)

wall = box(
    pos = vec(0,0,0),
    length = 10,
    width = 10,
    hieght = 10
)


# tiempo de colision
tmax = 2*A/B
nmax = 700
t = np.linspace(0, tmax, nmax)
dt = t[1]-t[0]

v_rocket = 0.0
y_rocket = 0.0

v_rock = 0.0
y_rock = s.pos.y

g1 = graph( width=500, height=0.4*500, align='right', 
            xtitle='time (s)', ytitle='y (m)' )
pos_rock = gcurve(label = "y rock", legend = True, graph=g1, color = color.blue)
pos_rocket = gcurve(label = "y rocket", legend=True, graph=g1, color = color.red)

g2 = graph( width=500, height=0.4*500, align='right',
            xtitle='time (s)', ytitle='velocity' )
vel_rock = gcurve(label = "v rock",legend = True, graph=g2, color = color.blue)
vel_rocket = gcurve(label = "v rocket", legend = True, graph=g2, color= color.red)


for ti in t:
    rate(100)
    
    v_rocket = v_rocket + a(ti)*dt
    y_rocket = y_rocket + v_rocket * dt
    rocket.pos.y = y_rocket
    
    v_rock = v_rock + _a(ti)*dt
    y_rock = y_rock + v_rock*dt
    s.pos.y = y_rock

    pos_rock.plot(ti, y_rock)
    pos_rocket.plot(ti, y_rocket)

    vel_rock.plot(ti,v_rock)
    vel_rocket.plot(ti,v_rocket)
    
```
