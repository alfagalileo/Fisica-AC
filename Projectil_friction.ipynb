```python
from vpython import *
import numpy as np
```

```python
g = 9.8

def ax(vx, t):
    return -k*vx

def ay(vy, t):
    return -k*vy - g

c = canvas(
    align="Center",
    width = 1000,
    height = 500,
    title='Simulación de cinemática',
    background = color.white,
    range = 16
)


ball = sphere(
    radius= 0.5,
    make_trail=True,
    pos = vec(0,0,0),
    color = color.blue
)

wall = box(
    pos=vec(0,0,0),
    length=50, height=1, width=10
)

k = 2.0
theta = 50
v0 = 15.0
vx = v0*np.cos( np.deg2rad(theta) )
vy = v0*np.sin( np.deg2rad(theta) )
x = 0
y = 0
t = 0

dt = 1e-3


g1 = graph( width=500, height=0.4*500, align='right', xtitle='x (m)', ytitle='y (m)' )
position = gcurve(graph=g1)

g2 = graph( width=500, height=0.4*500, align='left', xtitle='time (s)', ytitle='velocity')
velocity_x = gcurve(label = "vx",legend = True, graph=g2, color = color.blue)
velocity_y = gcurve(label = "vy", legend = True, graph=g2, color= color.red)

while True:
    rate(300)

    vx = vx + ax(t,vx)*dt
    x = x + vx*dt
    ball.pos.x = x 

    vy = vy + ay(t,vx)*dt
    y = y + vy*dt
    ball.pos.y = y 

    position.plot(x,y)
    velocity_x.plot(t,vx)
    velocity_y.plot(t,vy)
    
    t += dt
    
    if y < 0:
        break
```
