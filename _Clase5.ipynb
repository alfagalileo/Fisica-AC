```python
from vpython import *
```

```python
g, mu, dt = 9.8, 0.5, 0.01
theta = 25          
s, v = 12, 0 

scene = canvas(title = "<h1> Simulación bloque con fricción</h1>", width=600, height=500, background=color.white, range=16, align="left")
scene.userspin = False
scene.userzoom = False

tabla  = box(pos=vec(0, 0, 0), length=30, height=1, width=5, texture=textures.stucco)
objeto = box(length=4, height=1, width=1, color=color.blue)


def dibujar():
    th = radians(theta)
    u = vec( cos(th), sin(th), 0)
    n = vec( -sin(th), cos(th), 0 )
    tabla.axis = 30*u
    objeto.axis = 4*u
    objeto.pos = s*u + 1*n


def inclinacion(sl):
    global theta 
    theta = sl.value

def friccion(sl):
    global mu
    mu = sl.value


slider(bind=inclinacion, min=0, max=60, step=1, value=theta)
et_ang = wtext(text='  θ = {:2.0f}°     '.format(theta))
scene.append_to_caption('\n\n')

slider( bind=friccion, min=0, max=1, step=0.01, value=mu )
et_mu = wtext(text='  μ = {:.2f}   (θ crítico = {:.1f}°)'.format(mu, degrees(atan(mu))))

g1 = graph( width=500, height=300, align='right', xtitle='time (s)', ytitle='velocity' )
velocity = gcurve(label = "v",legend = True, graph=g1, color = color.blue)

g2 = graph( width=500, height=300, align='right', xtitle='time (s)', ytitle='position' )
position_x = gcurve(label = "x",legend = True, graph=g2, color = color.red)
position_y = gcurve(label = "y",legend = True, graph=g2, color = color.green)

t = 0.0
while True:
    rate(100)
    dibujar()
```
