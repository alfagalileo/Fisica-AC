```python
from vpython import (canvas, box, helix, graph, gcurve, vec, color, rate,
                     pi, sqrt, cos, sin)

# 1. Parámetros físicos
m     = 1.0                # masa del bloque [kg]
k     = 4.0 * pi**2        # constante elástica [N/m]; con m = 1 da omega = 2*pi
A     = 0.8                # amplitud inicial [m]
omega = sqrt(k / m)        # frecuencia angular [rad/s]
T     = 2.0 * pi / omega   # periodo [s]


L0       = 2.0                          # longitud natural del resorte [m]
lado     = 0.4                          # arista del bloque [m]
x_eq     = 0.0                          # posición de equilibrio del bloque
x_muro   = x_eq - lado / 2 - L0         # cara interna del muro
y_mesa   = -1.0                         # altura del centro de la mesa
e_mesa   = 0.1                          # espesor de la mesa
y_bloque = y_mesa + e_mesa / 2 + lado / 2 # el bloque descansa sobre la mesa


# 3. Escena 3D
scene = canvas(title="<h1>Sistema masa-resorte</h1>", width=800, height=600, background=color.gray(0.15))

scene.autoscale = False
scene.range     = 2.6

muro = box(pos=vec(x_muro, y_bloque, 0), size=vec(0.2, 1.2, 1.2), color=color.gray(0.6))
mesa = box(pos=vec(x_eq, y_mesa, 0), size=vec(5.0, e_mesa, 1.2), color=color.blue)
bloque = box(pos=vec(x_eq + A, y_bloque, 0), size=vec(lado, lado, lado), color=color.red)
resorte = helix(pos=vec(x_muro, y_bloque, 0), axis=vec(L0 + A, 0, 0), radius=0.15, thickness=0.03, coils=14, color=color.orange)


# 4. Gráficas
t_max = 4.0 * T

g_pos = graph(title="Posición vs tiempo", xtitle="t (s)", ytitle="x (m)",
              xmin=0, xmax=t_max, fast=False, width=800, height=300)
c_num = gcurve(graph=g_pos, color=color.red,  label="numérica")
c_ana = gcurve(graph=g_pos, color=color.black, label="analítica", dot=False)

g_vel = graph(title="Velocidad vs tiempo", xtitle="t (s)", ytitle="v (m/s)",
              xmin=0, xmax=t_max, fast=False, width=800, height=300)
c_vel = gcurve(graph=g_vel, color=color.blue, label="v(t)")


x  = A            # elongación respecto al equilibrio [m]
v  = 0.0          # velocidad [m/s]
t  = 0.0
dt = 1e-3         # paso de tiempo [s]

while t < t_max:
    rate(100)
    #TODO

    t += dt
```
